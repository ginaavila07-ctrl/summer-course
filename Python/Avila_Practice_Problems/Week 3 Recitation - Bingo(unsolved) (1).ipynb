# Number Bingo — live coding


## Rules
- The board is a text file: 4 lines, 4 numbers per line. Numbers range from 1 to 40 and may repeat.
- One person calls a clue from the menu. The other runs the game and enters its code.
- Mark every match with X. Earlier marks stay. Clues may repeat.
- A complete **row/column** wins. Choose Q to quit. AN marks every number.

Columns are an extension for you to try.


## 1. Helpers
Solutions from the warm-up.


```python
def is_even(n):
    return n % 2 == 0


def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, n):
        if n % divisor == 0:
            return False
    return True


def is_multiple_of(n, divisor):
    return n % divisor == 0


def display_file(file_name):
    print('-' * 25)
    with open(file_name) as f:
        for line in f:
            print(line, end='')
    print('-' * 25)


def display_options():
    print()
    print(' E : Even')
    print(' P : Prime')
    print(' M1: Multiple of 3')
    print(' M2: Multiple of 5')
    print(' L1: Less than 15')
    print(' G1: Greater than 25')
    print(' AN: Any number')
    print(' Q : Quit')


def get_user_choice():
    valid_choices = 'E P M1 M2 L1 G1 AN Q'
    display_options()
    choice = input('Choose a clue: ').strip().upper()
    while choice not in valid_choices.split():
        print('Invalid option. Try again.')
        choice = input('Choose a clue: ').strip().upper()
    return choice
```

## 3. Mark one value

This function decides to keep the given cell_value as is or to replace iwth 'X' as a mark. For example
- `mark_for(7, 'P')` returns `'X'`.
- `mark_for('7', 'P')` returns `'X'`.
- `mark_for('12', 'P')` returns `'12'`.
- An existing `'X'` stays `'X'` -> mark_for('X', ____) returns 'X'


```python
def mark_for(cell_value, clue):
    # TODO: keep an existing 'X' as is.
    # Convert cell_value using the relevant helper functions you developed.
    # Return 'X' if it matches the clue code (e.g. if it is an even number for 'E').
    # Otherwise, return the number itself as a string.
    number = int(cell_value)
    if cell_value == 'X':
        return "X"

    

    elif clue == 'E' and is_even(number):
        return 'X'
    elif clue == 'P' and is_prime(number):
        return 'X'
    elif clue == 'M1' and is_multiple_of(number, 3):
        return 'X'
    elif clue == 'M2' and is_multiple_of(number, 5):
        return 'X'
    elif clue == 'L1' and number < 15:
        return 'X'
    elif clue ==  'G1' and number > 25:
        return 'X'
    elif clue == "AN" and isinstance(number, int):
        return 'X'
    else: 
        return str(number)
```

## 4. Update the file
Goal is to read the board and mark 'X's for the matching cell_values.

Loop through the lines, then extract the cell values from `line.split()`.

Build the new text as the file content before opening the same file for writing.


```python
def mark_matches(board_file, clue):
    # TODO 1: Make necessary changes to mark each cell value
    # -> generate out_string using the mark_for function
    # -> Use f'{______:>5}' to keep the columns aligned.

    out_string = ''

    # open file for reading to get board data to update
    with open(board_file) as f:

        for line in f:

            for item in line.split():
                out_string = out_string + f'{mark_for(item, clue):>5}'

            out_string = out_string + '\n'


    # TODO 2: open the file for writing and save your out_string.
    with open(board_file, 'w') as f:

        f.write(out_string)
```

## 5. Check for bingo
Check only the rows to see if it is bingo.


```python
def check_row_bingo(file_name):
    # TODO: read one line at a time.
    
    with open(file_name, 'r') as f_board:
        for row in f_board:
            
            if row.count('X') == 4:
                return True
            return False

    # If all 4 cell values in a row are 'X' return True (it is bingo)
    # If there are no rows as bingo return False
    
```

## 6. Play
The game loop is provided. Run this after completing the functions.


```python
def play_game(board_file):

    display_file(board_file)
    while not check_row_bingo(board_file):
        clue = get_user_choice()
        if clue == 'Q':
            print('Game stopped.')
            return
        mark_matches(board_file, clue)
        display_file(board_file)
    print('BINGO! A row is complete.')
```
